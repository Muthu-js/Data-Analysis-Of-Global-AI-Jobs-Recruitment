#**Dataset Information**

**Source:** Open Data Platforms / Kaggle (https://www.kaggle.com/datasets/datascikhan/global-ai-jobs-2026?select=global_ai_jobs_2026.csv)

**Year / Timeline:** Data collected during 2026

**Dataset Column/features Description:**

Job_ID – Unique identification number assigned to each AI job posting.

Job_Title – Name or designation of the job position.

AI_Specialization – Area of AI associated with the job, such as Computer Vision, Generative AI, or Reinforcement Learning.

Industry – Industry or business sector in which the job opportunity is available.

Company – Name of the company offering the job position.

Country – Country where the job opportunity is located.

Continent – Continent or geographical region associated with the job location.

Education_Required – Minimum educational qualification required for the position.

Years_of_Experience – Number of years of professional experience required for the job.

Salary_Category – Classification of the salary offered, such as Low, Medium, High, or Very High.

Skills_Required – Technical and professional skills required for the job, such as Python, SQL, Power BI, PyTorch, and Tableau.

Required_Certifications – Certifications required or preferred for the position, such as AWS Certified or DeepLearning.ai.

Job_Openings – Number of available positions or openings for the particular job role.

Job_Type – Type of employment, such as Full-time, Part-time, or Contract.

## Stage 3 – EDA and Visualizations

Univariate Analysis → distribution of single variables (countplot, histogram, boxplot)

Bivariate Analysis → relation between two variables (scatterplot, barplot, correlation heatmap)

Multivariate Analysis → relation among 3+ variables (pairplot, grouped analysis, pivot tables, advanced plots)

Focus on business story not just charts

```python
# # Univariate analysis
# Countplot - Job Type

sns.set_theme(style="whitegrid")

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x='Job Type',
    order=df['Job Type'].value_counts().index
)

plt.title("Distribution of AI Jobs by Job Type")
plt.xlabel("Job Type")
plt.ylabel("Number of Job Records")
#plt.xticks(rotation=50)
plt.tight_layout()
plt.show()

# Interpretation
job_counts = df['Job Type'].value_counts()

print("INTERPRETATION:")
print("1. The chart shows the distribution of job records by employment type.")
print(f"2. Most common job type: {job_counts.index[0]}")
print(f"3. Number of records: {job_counts.iloc[0]}")
print(f"4. Least common job type: {job_counts.index[-1]}")
print(f"5. Number of records: {job_counts.iloc[-1]}")

print("\nBUSINESS INSIGHT:")
print("This analysis helps job seekers understand which employment "
      "types are most represented in the dataset.")
```

```python
#Countplot of AI Specialization

sns.set_theme(style="whitegrid")

plt.figure(figsize=(11, 6))

sns.countplot(
    data=df,
    y='AI Specialization',
    order=df['AI Specialization'].value_counts().index
)

plt.title("Distribution of AI Jobs by Specialization")
plt.xlabel("Number of Job Records")
plt.ylabel("AI Specialization")
plt.tight_layout()
plt.show()

# Interpretation
specialization_counts = df['AI Specialization'].value_counts()

print("INTERPRETATION:")
print("1. The chart shows the frequency of each AI specialization.")
print(f"2. Most frequent specialization: {specialization_counts.index[0]}")
print(f"3. Number of records: {specialization_counts.iloc[0]}")
print(f"4. Least frequent specialization: {specialization_counts.index[-1]}")
print(f"5. Number of records: {specialization_counts.iloc[-1]}")

print("\nBUSINESS INSIGHT:")
print("This analysis helps job seekers identify the most represented "
      "AI specializations and plan their skill development.")
```

```python
# Histogram - Years of Experience

plt.figure(figsize=(9, 5))
sns.histplot(data=df, x='Years of Experience', bins=15, kde=True) # Kernel Density Estimate (KDE)
plt.title("Distribution of Required Years of Experience")
plt.xlabel("Years of Experience")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("INTERPRETATION:")
print(f"1. Mean experience: {df['Years of Experience'].mean():.2f} years.")
print(f"2. Median experience: {df['Years of Experience'].median():.2f} years.")
print("3. The histogram shows whether experience requirements are "
      "concentrated at lower, middle, or higher values.")
print("4. BUSINESS INSIGHT: This helps candidates compare their "
      "experience with the roles represented in the dataset.")
```

```python
# Histogram - Job Openings

plt.figure(figsize=(9, 5))
sns.histplot(data=df, x='Job Openings', bins=15, kde=True)

plt.title("Distribution of Job Openings")
plt.xlabel("Number of Job Openings")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()

print("INTERPRETATION:")
print(f"1. Average openings per record: {df['Job Openings'].mean():.2f}.")
print(f"2. Minimum openings: {df['Job Openings'].min()}.")
print(f"3. Maximum openings: {df['Job Openings'].max()}.")
print("4. BUSINESS INSIGHT: This chart shows whether most job records "
      "have a small or large number of openings.")
```

```python
# Boxplot - Experience Outliers

plt.figure(figsize=(9, 4))
sns.boxplot(data=df, x='Years of Experience')

plt.title("Boxplot of Years of Experience")
plt.xlabel("Years of Experience")
plt.tight_layout()
plt.show()

q1 = df['Years of Experience'].quantile(0.25)
q3 = df['Years of Experience'].quantile(0.75)
iqr = q3 - q1

outliers = df[
    (df['Years of Experience'] < q1 - 1.5 * iqr) |
    (df['Years of Experience'] > q3 + 1.5 * iqr)
]

print("INTERPRETATION:")
print(f"1. Number of potential outliers: {len(outliers)}.")
print("2. Points outside the whiskers may indicate unusually high "
      "or low experience requirements.")
print("3. BUSINESS INSIGHT: Check these records for errors or "
      "specialized senior roles before changing or removing values.")
```

```python
## Bivariate analysis
#Bivariate analysis examines the relationship between two variables.

# Scatterplot - Experience and Job Openings

plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x='Years of Experience',
    y='Job Openings',
    alpha=0.5
)

plt.title("Years of Experience vs Job Openings")
plt.xlabel("Years of Experience")
plt.ylabel("Job Openings")
plt.tight_layout()
plt.show()

correlation = df['Years of Experience'].corr(df['Job Openings'])

print("INTERPRETATION:")
print(f"1. Correlation coefficient: {correlation:.3f}.")
print("2. A positive value indicates a positive linear relationship; "
      "a negative value indicates a negative linear relationship.")
print("3. Values near zero indicate little linear association.")
print("4. BUSINESS INSIGHT: Experience requirements alone may not "
      "explain the number of openings.")
```

```python
# Barplot - Industry and Job Openings

industry_openings = (
    df.groupby('Industry')['Job Openings']
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))
sns.barplot(x=industry_openings.values, y=industry_openings.index)

plt.title("Top 10 Industries by Average Job Openings")
plt.xlabel("Average Job Openings")
plt.ylabel("Industry")
plt.tight_layout()
plt.show()

print("INTERPRETATION:")
print(f"1. Highest average among the displayed industries: "
      f"{industry_openings.index[0]}.")
print(f"2. Average openings in that industry: "
      f"{industry_openings.iloc[0]:.2f}.")
print("3. BUSINESS INSIGHT: Compare industries to identify where "
      "individual job records to have more openings.")
```

```python
# Correlation Heatmap

numeric_columns = ['Years of Experience', 'Job Openings']
correlation_matrix = df[numeric_columns].corr()

plt.figure(figsize=(7, 5))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

print("INTERPRETATION:")
print(correlation_matrix.round(3))
print("1. Values close to +1 indicate strong positive linear association.")
print("2. Values close to -1 indicate strong negative linear association.")
print("3. Values close to 0 indicate weak linear association.")
print("4. BUSINESS INSIGHT: Correlation helps identify numerical "
      "relationships but does not establish cause and effect.")
```

```python
## Multivariate analysis

# Pairplot - Numerical Variables

sns.pairplot(
    df[['Years of Experience', 'Job Openings']],
    diag_kind='hist',
    plot_kws={'alpha': 0.5}
)

plt.show()

print("INTERPRETATION:")
print("1. The diagonal plots show the distribution of each numerical variable.")
print("2. The off-diagonal plots show the relationship between the variables.")
print("3. BUSINESS INSIGHT: This helps identify patterns and unusual "
      "records that may need further investigation.")
```

```python
# Continent, job type, and job openings

# Grouped Analysis

grouped_data = (
    df.groupby(['Country', 'Job Type'])['Job Openings']
      .sum()
      .reset_index()
)

plt.figure(figsize=(20, 10))
sns.barplot(
    data=grouped_data,
    x='Country',
    y='Job Openings',
    hue='Job Type',
    errorbar=None
)

plt.title("Total Job Openings by Country and Job Type")
plt.xlabel("Country")
plt.ylabel("Total Job Openings")
plt.xticks(rotation=70)
plt.legend(title="Job Type", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()

print("INTERPRETATION:")
print("1. The chart compares total openings across Country and job types.")
print("2. Compare the heights of bars within and across Country.")
print("3. BUSINESS INSIGHT: This can help recruiters compare the "
      "geographical distribution of different employment types.")
```

```python
# Pivot table — industry and specialization

pivot_table = pd.pivot_table(
    df,
    index='Industry',
    columns='AI Specialization',
    values='Job Openings',
    aggfunc='sum',
    fill_value=0
)

display(pivot_table)

print("INTERPRETATION:")
print("1. Rows represent industries.")
print("2. Columns represent AI specializations.")
print("3. Each cell shows the total recorded job openings for "
      "that industry-specialization combination.")
print("4. BUSINESS INSIGHT: This helps identify combinations "
      "with relatively high recorded hiring volume.")
```

```python
# Education versus salary category

salary_education = pd.crosstab(
    df['Education Required'],
    df['Salary Category']
)

display(salary_education)

salary_education.plot(
    kind='bar',
    stacked=True,
    figsize=(11, 6)
)

plt.title("Salary Categories by Education Requirement")
plt.xlabel("Education Required")
plt.ylabel("Number of Job Records")
plt.xticks(rotation=25)
plt.legend(title="Salary Category", bbox_to_anchor=(1.02, 1))
plt.tight_layout()
plt.show()

print("INTERPRETATION:")
print("1. Each bar represents an education requirement.")
print("2. The colored sections show the distribution of salary categories.")
print("3. BUSINESS INSIGHT: This helps explore how salary categories "
      "vary across education requirements.")
print("4. LIMITATION: Salary categories are not actual salary amounts.")
```

# Types of Analysis

## 1. Descriptive Analysis
Descriptive analysis summarizes the Global AI Jobs 2026 dataset by examining AI specializations, industries, countries, job types, salary categories, education requirements, and experience levels. Charts and statistical summaries help identify the most common job categories and the distribution of job openings.

## 2. Diagnostic Analysis
Diagnostic analysis investigates patterns and relationships among job characteristics. It compares job openings across industries and locations, examines salary categories across AI specializations, and explores the relationship between required experience and job openings. These findings help identify patterns that may explain differences in advertised hiring demand.

## 3. Predictive Analysis (Optional)
Predictive analysis uses machine learning to estimate outcomes, such as salary categories, based on job characteristics. A classification model can be trained and evaluated using training and testing datasets. Since historical hiring data is not available, reliable time-based forecasting of future job demand would require additional historical data.

## 4. Prescriptive Analysis
Prescriptive analysis converts the findings into actionable recommendations. These may include prioritizing relevant AI skills for training, identifying industries and locations for further recruitment research, and evaluating education and experience requirements. Recommendations should be based on the observed data and validated against real-world employment conditions.

## Expected Outcome
The analysis aims to provide a clearer understanding of global AI job opportunities, the characteristics associated with job openings, and the potential implications for job seekers, recruiters, and training providers. The results can support data-driven decisions about skill development, recruitment priorities, and career planning.
